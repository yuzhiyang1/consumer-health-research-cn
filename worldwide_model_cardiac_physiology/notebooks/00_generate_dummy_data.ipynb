# 00——合成输入数据(`data/dummy_data*.csv`).

Fitbit 组别不能重新分配，这个笔记本写着两个小的**合成**
公共输入计划中的用户月表，以便`01_cohort_stats.ipynb`,
`02_modelling.ipynb`和`03_model_results.ipynb`可以运行到结束：

| 文件 | 度量衡 | 列数 |
| :--- | :--- | :--- |
| `dummy_data.csv` | 静息心率 | `id, month, age, sex, country, latitude, longitude, temperature, temperature min, temperature max, day light, resting heart rate, region` |
| `dummy_data_hrv.csv` | 心率变异性 | 同，以`heart rate variability`为目标。 |

运行全部单元格一次。文件被写入此 仓库 旁边的数据根( `./data` ) ,
或 `$WPM_DATA_DIR`; 现有文件未受影响，除非`OVERWRITE = True`.
**从这些文件中估计到的任何内容都无意义** ——它们只行使代码路径。
为了分析你自己的数据，将其写入同一个计划(见`data/README.md`)并跳过
这本笔记本。

```python
# ----------------------------------------------------------------------------
# CONFIG
# ----------------------------------------------------------------------------
N_USERS = 300  # RHR用户；HRV组是一个~2/3子集
SEED = 0  # 固定种子 ->每次运行时相同的文件
OVERWRITE = False  # True 重生已有文件

data_dir = fio.get_data_dir()
print('data root:', data_dir)
```

```python
targets = {
    m: os.path.join(data_dir, cio.PUBLIC_CSV_NAMES[m]) for m in ('rhr', 'hrv')
}
existing = [p for p in targets.values() if fio.file_exists(p)]
if existing and not OVERWRITE:
  print('already present (set OVERWRITE = True to regenerate):')
  for p in existing:
    print('  ', p)
else:
  written = gdd.write_dummy_data(data_dir, n_users=N_USERS, seed=SEED)
```

```python
# 快看:通过分析笔记本所用的同一加载器进行计程检查。
for metric in ('rhr', 'hrv'):
  df = cio.load_cohort(data_dir, metric)
  print(
      f'{metric}: {len(df):,} rows, {df["user_id"].nunique():,} users, '
      f'{df["activity_period"].min()} .. {df["activity_period"].max()}'
  )
df.head()
```
